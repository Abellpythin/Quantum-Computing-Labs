```python
import qiskit
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.draw(output = "mpl")
#B00 = 1 √2 (|00⟩ + |11⟩)
# Below is an explaination 
```

'''What the circuit does step-by-step:
Start with |00⟩: Both qubits begin in the ground state |0⟩

qc.h(0) - Hadamard on qubit 0:

Creates superposition on qubit 0
The Hadamard transforms |0⟩ → 1/√2(|0⟩ + |1⟩)
Combined state becomes: 1/√2(|00⟩ + |10⟩)
Qubit 0 is now in equal superposition, qubit 1 remains |0⟩



qc.cx(0,1) - CNOT gate (Controlled-NOT):

Qubit 0 is the control, qubit 1 is the target
Creates entanglement between the two qubits
When qubit 0 is |0⟩: |00⟩ stays |00⟩ (no flip)
When qubit 0 is |1⟩: |10⟩ becomes |11⟩ (qubit 1 flips)
Final state: 1/√2(|00⟩ + |11⟩)


Result: This circuit creates the Φ⁺ Bell state (B00) = 1/√2(|00⟩ + |11⟩)'''

```python
import qiskit
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.x(1)

qc.draw(output = "mpl")

#B01 = 1 √2 (|01⟩ + |10⟩) 
# Below is an explaination 
```

'''What the circuit does step-by-step:

Start with |00⟩: Both qubits begin in the ground state |0⟩
qc.h(0) - Hadamard on qubit 0:

Creates superposition on qubit 0
Transforms |00⟩ → 1/√2(|00⟩ + |10⟩)
Qubit 0 is now in equal superposition of |0⟩ and |1⟩


qc.cx(0,1) - CNOT gate:

Qubit 0 is the control, qubit 1 is the target
Creates entanglement between the two qubits
When qubit 0 is |0⟩: |00⟩ stays |00⟩
When qubit 0 is |1⟩: |10⟩ flips to |11⟩
State becomes: 1/√2(|00⟩ + |11⟩)
This is the Φ⁺ Bell state


qc.x(1) - X gate (NOT gate) on qubit 1:

Flips qubit 1 in both terms of the superposition
|00⟩ → |01⟩ (qubit 1 flips from |0⟩ to |1⟩)
|11⟩ → |10⟩ (qubit 1 flips from |1⟩ to |0⟩)
Final state: 1/√2(|01⟩ + |10⟩)'''

Adds a negative phase to the |1⟩ component of qubit 0
Transforms 1/√2(|00⟩ + |11⟩) → 1/√2(|00⟩ - |11⟩)

```python
import qiskit
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0,1)
qc.z(1)
qc.draw(output = "mpl")
# B10= 1 √2 (|00⟩ − |11⟩) - if qc.z(1) is symetric to qc.z(0)
# Below is an explaination 
```

'''What the circuit does:

```python
import qiskit
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.h(0)        # Hadamard on qubit 0
qc.cx(0, 1)    # CNOT
qc.x(1)        # X gate on qubit 1
qc.z(0)        # Z gate on qubit 0
qc.draw(output = "mpl")
```

'''What the circuit does step-by-step:

Start with |00⟩: Both qubits begin in the ground state |0⟩
qc.h(0) - Hadamard on qubit 0:

Creates superposition on qubit 0
Transforms |00⟩ → 1/√2(|00⟩ + |10⟩)
Qubit 0 is now in equal superposition of |0⟩ and |1⟩


qc.cx(0,1) - CNOT gate:

Qubit 0 is the control, qubit 1 is the target
Creates entanglement
When qubit 0 is |0⟩: |00⟩ stays |00⟩
When qubit 0 is |1⟩: |10⟩ becomes |11⟩
State becomes: 1/√2(|00⟩ + |11⟩)
This is the Φ⁺ Bell state


qc.x(1) - X gate (NOT gate) on qubit 1:

Flips qubit 1 in both terms of the superposition
|00⟩ → |01⟩ (qubit 1 flips from |0⟩ to |1⟩)
|11⟩ → |10⟩ (qubit 1 flips from |1⟩ to |0⟩)
State becomes: 1/√2(|01⟩ + |10⟩)


qc.z(0) - Z gate on qubit 0:

Adds a negative phase when qubit 0 is in state |1⟩
The |01⟩ term is unchanged (qubit 0 is |0⟩)
The |10⟩ term gets a minus sign (qubit 0 is |1⟩)
Final state: 1/√2(|01⟩ - |10⟩)



Result: This circuit creates the Ψ⁻ Bell state (B11) = 1/√2(|01⟩ - |10⟩)'''

```python
import qiskit
from qiskit import QuantumCircuit
qc = QuantumCircuit(2)
qc.x(1)        # X gate on qubit 1
qc.h(0)        # Hadamard on qubit 0
qc.z(0)        # Z gate on qubit 0
qc.cx(0, 1)    # CNOT
qc.draw(output = "mpl")

#B11 = 1 √2 (|01⟩ − |10⟩)
# Below is an explaination 
```

'''What the circuit does step-by-step:

Start with |00⟩: Both qubits begin in the ground state |0⟩
qc.x(1) - X gate (NOT gate) on qubit 1:

Flips qubit 1 from |0⟩ to |1⟩
State becomes |01⟩
Qubit 0 is still |0⟩, qubit 1 is now |1⟩


qc.h(0) - Hadamard on qubit 0:

Creates superposition on qubit 0
Hadamard on |0⟩ gives: 1/√2(|0⟩ + |1⟩)
Combined state becomes: 1/√2(|01⟩ + |11⟩)
Qubit 1 remains in state |1⟩ throughout


qc.z(0) - Z gate on qubit 0:

Adds a negative phase when qubit 0 is in state |1⟩
The |11⟩ term gets a minus sign (since qubit 0 is |1⟩ there)
The |01⟩ term is unchanged (since qubit 0 is |0⟩ there)
State becomes: 1/√2(|01⟩ - |11⟩)


qc.cx(0,1) - CNOT gate:

Qubit 0 is the control, qubit 1 is the target
Creates entanglement by flipping qubit 1 when qubit 0 is |1⟩
When qubit 0 is |0⟩: |01⟩ stays |01⟩
When qubit 0 is |1⟩: |11⟩ flips to |10⟩
Final state: 1/√2(|01⟩ - |10⟩) '''
